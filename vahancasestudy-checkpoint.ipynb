```python
import pandas as pd

df = pd.read_csv("Vahan_Case_Study (1).csv")

print(df.shape)
```

```python

model_df = df.copy()

model_df["upload_date"] = pd.to_datetime(model_df["upload_date"])

model_df["upload_dayofweek"] = model_df["upload_date"].dt.dayofweek
model_df["upload_day"] = model_df["upload_date"].dt.day

drop_cols = [
    "candidate_phone",
    "upload_date",
    "FT_after_first_attempt",
    "Attempted → FT_after_upload %",
    "Interested → FT_after_first_attempt %"
]

model_df = model_df.drop(columns=drop_cols)

print(model_df.shape)
model_df.head()
```

```python
print("Target distribution:")
print(model_df["FT_after_upload"].value_counts())

print("\nTarget percentage:")
print(model_df["FT_after_upload"].value_counts(normalize=True) * 100)

print("\nMissing values:")
print(model_df.isnull().sum()[model_df.isnull().sum() > 0])
```

```python
from sklearn.model_selection import train_test_split

# Target
y = model_df["FT_after_upload"]

# Features
X = model_df.drop(columns=["FT_after_upload"])

# Split the data while preserving the rare FT cases
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training set:", X_train.shape)
print("Test set:", X_test.shape)

print("\nFT cases in training:", y_train.sum())
print("FT cases in test:", y_test.sum())
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

categorical_features = ["lead_source"]

preprocessor = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ],
    remainder="passthrough"
)

X_train_encoded = preprocessor.fit_transform(X_train)
X_test_encoded = preprocessor.transform(X_test)

print("Encoded training shape:", X_train_encoded.shape)
print("Encoded test shape:", X_test_encoded.shape)
```

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(
    n_estimators=500,
    random_state=42,
    class_weight="balanced",
    min_samples_leaf=5,
    n_jobs=-1
)

model.fit(X_train_encoded, y_train)

print("Model trained successfully!")
```

```python
from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    roc_auc_score
)

y_pred = model.predict(X_test_encoded)
y_prob = model.predict_proba(X_test_encoded)[:, 1]

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))

print("\nROC-AUC:", roc_auc_score(y_test, y_prob))
```

```python
import pandas as pd

feature_names = preprocessor.get_feature_names_out()

importance_df = pd.DataFrame({
    "Feature": feature_names,
    "Importance": model.feature_importances_
}).sort_values("Importance", ascending=False)

importance_df.head(15)
```

```python
# Conservative feature set to reduce potential target leakage

safe_features = [
    "lead_source",
    "Uploaded Leads",
    "Attempted",
    "Connected",
    "Attempt per Lead",
    "tag_filled",
    "Interested",
    "upload_day",
    "upload_dayofweek"
]

X_safe = model_df[safe_features]
y_safe = model_df["FT_after_upload"]

X_train_s, X_test_s, y_train_s, y_test_s = train_test_split(
    X_safe,
    y_safe,
    test_size=0.20,
    random_state=42,
    stratify=y_safe
)

categorical_features = ["lead_source"]

preprocessor_safe = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ],
    remainder="passthrough"
)

X_train_s_enc = preprocessor_safe.fit_transform(X_train_s)
X_test_s_enc = preprocessor_safe.transform(X_test_s)

safe_model = RandomForestClassifier(
    n_estimators=500,
    random_state=42,
    class_weight="balanced",
    min_samples_leaf=5,
    n_jobs=-1
)

safe_model.fit(X_train_s_enc, y_train_s)

y_pred_s = safe_model.predict(X_test_s_enc)
y_prob_s = safe_model.predict_proba(X_test_s_enc)[:, 1]

print("Confusion Matrix:")
print(confusion_matrix(y_test_s, y_pred_s))

print("\nClassification Report:")
print(classification_report(y_test_s, y_pred_s, zero_division=0))

print("\nROC-AUC:", roc_auc_score(y_test_s, y_prob_s))
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score

thresholds = [0.5, 0.4, 0.3, 0.2, 0.1]

for threshold in thresholds:
    pred = (y_prob_s >= threshold).astype(int)
    
    print(f"\nThreshold: {threshold}")
    print("Precision:", round(precision_score(y_test_s, pred, zero_division=0), 3))
    print("Recall:", round(recall_score(y_test_s, pred, zero_division=0), 3))
    print("F1:", round(f1_score(y_test_s, pred, zero_division=0), 3))
    print("Confusion Matrix:")
    print(confusion_matrix(y_test_s, pred))
```
